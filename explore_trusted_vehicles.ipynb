# Trusted vehicles explorer

Load and inspect the trusted `vehicles` parquet dataset.

```python
import os
from pathlib import Path

from pyspark.sql import SparkSession, functions as F

DATA_DIR = Path(os.environ.get("DATA_DIR", "data")).expanduser().resolve()
VEHICLES_PATH = DATA_DIR / "trusted" / "vehicles"

spark = SparkSession.builder.appName("trusted-vehicles-explorer").getOrCreate()

print("DATA_DIR:", DATA_DIR)
print("VEHICLES_PATH:", VEHICLES_PATH)
```

```python
df = spark.read.parquet(str(VEHICLES_PATH))

df.printSchema()
print("rows:", df.count())
df.show(20, truncate=False)
```

```python
df.groupBy("year", "month", "day").agg(
    F.min("event_timestamp").alias("min_ts"),
    F.max("event_timestamp").alias("max_ts"),
    F.countDistinct("vehicle").alias("vehicles"),
    F.count("*").alias("rows"),
).orderBy("year", "month", "day").show(50, truncate=False)

df.select(
    F.min("latitude").alias("min_latitude"),
    F.max("latitude").alias("max_latitude"),
    F.min("longitude").alias("min_longitude"),
    F.max("longitude").alias("max_longitude"),
).show(truncate=False)
```
